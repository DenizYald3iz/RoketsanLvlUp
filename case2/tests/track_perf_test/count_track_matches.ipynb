```python
import pandas as pd

# Change this to your actual CSV filename/path.
CSV_PATH = "./debug/track_matches.csv"

df = pd.read_csv(CSV_PATH)
print(f"Rows loaded: {len(df)}")
display(df.head())
```

```python
# Count how many times every track_id occurs.
track_counts = df["track_id"].value_counts()

# Keep only track IDs that occur at least twice.
duplicates = track_counts[track_counts > 1]

print(f"Unique track IDs: {df['track_id'].nunique()}")
print(f"Track IDs occurring more than once: {len(duplicates)}")

if duplicates.empty:
    print("No duplicated TXXXX values found.")
else:
    display(duplicates.rename("count").to_frame())
```

```python
# Show every row belonging to a duplicated track_id.
duplicate_rows = (
    df[df["track_id"].isin(duplicates.index)]
    .sort_values(["track_id", "time", "image_id"])
)

display(duplicate_rows)
```

```python
# Optional: print a compact text summary.
for track_id, count in duplicates.items():
    image_ids = df.loc[df["track_id"] == track_id, "image_id"].tolist()
    print(f"{track_id}: {count} occurrences -> {image_ids}")
```

```python
from pathlib import Path
import pandas as pd
import re

TRACK_SUMMARIES_DIR = Path("track_summaries")
CSV_PATH = "./debug/track_matches.csv"

# Load track IDs from CSV
df = pd.read_csv(CSV_PATH)
csv_track_ids = set(df["track_id"].astype(str))

# Get TXXXX.json files from folder
pattern = re.compile(r"^T\d{4}\.json$")

json_track_ids = {
    p.stem
    for p in TRACK_SUMMARIES_DIR.iterdir()
    if p.is_file() and pattern.match(p.name)
}

# Present in track_summaries, but absent from track_matches.csv
missing_from_csv = sorted(json_track_ids - csv_track_ids)

print(f"TXXXX.json files in track_summaries: {len(json_track_ids)}")
print(f"Unique TXXXX in track_matches.csv:   {len(csv_track_ids)}")
print(f"In track_summaries but NOT in CSV:   {len(missing_from_csv)}")
print()

for track_id in missing_from_csv:
    print(track_id)
```
